```python
def map_label(label):
    label = str(label)
    if label == 'BENIGN':
        return 'BENIGN'
    elif label.startswith('DDOS-'):
        return 'DDOS'
    elif label.startswith('DOS-'):
        return 'DOS'
    elif label.startswith('MIRAI-'):
        return 'MIRAI'
    elif label.startswith('RECON-'):
        return 'RECON'
    elif label in ['MITM-ARPSPOOFING', 'DNS_SPOOFING']:
        return 'MITM_SPOOFING'
    elif label in ['DICTIONARYBRUTEFORCE', 'COMMANDINJECTION', 'SQLINJECTION', 'XSS', 'UPLOADING_ATTACK']:
        return 'EXPLOIT'
    elif label in ['BROWSERHIJACKING', 'BACKDOOR_MALWARE']:
        return 'MALWARE'
    elif label == 'VULNERABILITYSCAN':
        return 'SCAN'
    else:
        return 'OTHER'

df['Grouped_Label'] = df['Label'].astype(str).apply(map_label)
```

```python
import numpy as np
print("Cleaning data...")
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
print(df.isnull().sum())
```

```python
import pandas as pd

# Assuming your DataFrame is named df
df = df.drop('Label', axis=1)
```

```python
df['Grouped_Label'].value_counts()
```

```python
import os
from imblearn.over_sampling import SMOTE
from collections import Counter

# Optional: Silence joblib warning
os.environ["LOKY_MAX_CPU_COUNT"] = "8"

# Step 2: Clean the data
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.fillna(df.median(numeric_only=True), inplace=True)

# Step 3: Separate features and target
X = df.drop('Grouped_Label', axis=1)
y = df['Grouped_Label']

# Step 4: Apply SMOTE to the entire dataset
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

# Step 5: Show class distributions
print("Before SMOTE:", Counter(y))
print("After SMOTE:", Counter(y_resampled))

# Step 6: Combine and save the balanced dataset
balanced_df = pd.concat([pd.DataFrame(X_resampled, columns=X.columns),
                         pd.Series(y_resampled, name='Label')], axis=1)

# Save to Desktop
output_path = r"C:\Users\rahul\Desktop\Dataset\balanced_dataset_5_9 class median.csv"
balanced_df.to_csv(output_path, index=False)
print(f"✅ Balanced dataset saved to Desktop as '{os.path.basename(output_path)}'")
```
